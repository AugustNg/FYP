```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import lightgbm as lgb
from sklearn.model_selection import train_test_split, cross_val_score, TimeSeriesSplit
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.pipeline import Pipeline
from bayes_opt import BayesianOptimization

# Load data
df = pd.read_csv('retail_store_inventory.csv')

# Feature engineering for units sold prediction
def create_sales_features(df):
    # Convert date
    df['Date'] = pd.to_datetime(df['Date'])
    df['day_of_week'] = df['Date'].dt.dayofweek
    df['day_of_month'] = df['Date'].dt.day
    df['is_weekend'] = (df['day_of_week'] >= 5).astype(int)
    df['is_month_start'] = (df['day_of_month'] == 1).astype(int)
    df['is_month_end'] = df['Date'].dt.is_month_end.astype(int)
    
    # Sort by store, product, and date
    df.sort_values(['Store ID', 'Product ID', 'Date'], inplace=True)
    
    # Lag features for units sold
    for lag in [1, 2, 3, 7, 14]:
        df[f'units_sold_lag_{lag}'] = df.groupby(['Store ID', 'Product ID'])['Units Sold'].shift(lag)
    
    # Rolling features (calculate after the train-test split to prevent data leakage)
    for window in [3, 7, 14]:
        df[f'sold_rolling_mean_{window}'] = df.groupby(['Store ID', 'Product ID'])['Units Sold'].transform(
            lambda x: x.rolling(window, min_periods=1).mean())
        df[f'sold_rolling_std_{window}'] = df.groupby(['Store ID', 'Product ID'])['Units Sold'].transform(
            lambda x: x.rolling(window, min_periods=1).std())
    
    # Price and discount features
    df['price_ratio'] = df['Price'] / df['Competitor Pricing']
    df['discount_effect'] = df['Discount'] * df['Price']
    
    # Inventory features
    df['inventory_ratio'] = df['Inventory Level'] / df['Units Sold'].replace(0, 1)
    df['stock_out_risk'] = (df['Inventory Level'] - df['Units Sold']) / df['Units Sold'].replace(0, 1)
    
    # Time since last sale
    df['days_since_last_sale'] = df.groupby(['Store ID', 'Product ID'])['Date'].diff().dt.days
    
    return df

df = create_sales_features(df)

# Define features and target - using Units Sold as target now
X = df.drop(['Units Sold', 'Date', 'Demand Forecast', 'Units Ordered'], axis=1)
y = df['Units Sold']

# Time-based train-test split (make sure no data leakage occurs)
train_size = int(0.8 * len(df))
X_train, X_test = X.iloc[:train_size], X.iloc[train_size:]
y_train, y_test = y.iloc[:train_size], y.iloc[train_size:]

# Preprocessing
numeric_features = ['Inventory Level', 'Price', 'Discount', 'Competitor Pricing',
                   'day_of_week', 'day_of_month', 'units_sold_lag_1', 
                   'units_sold_lag_2', 'units_sold_lag_3', 'units_sold_lag_7',
                   'units_sold_lag_14', 'sold_rolling_mean_3', 'sold_rolling_mean_7',
                   'sold_rolling_mean_14', 'sold_rolling_std_3', 'sold_rolling_std_7',
                   'price_ratio', 'discount_effect', 'inventory_ratio', 'stock_out_risk',
                   'days_since_last_sale']
categorical_features = ['Store ID', 'Product ID', 'Category', 'Region',
                       'Weather Condition', 'Holiday/Promotion', 'Seasonality',
                       'is_weekend', 'is_month_start', 'is_month_end']

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)])

# Function to evaluate the model performance
def evaluate_model(model, X_train, y_train, X_test, y_test):
    # Predict on the test set
    y_pred = model.predict(X_test)
    
    # Metrics
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    mae = mean_absolute_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    
    # Print metrics
    print(f"RMSE: {rmse:.4f}")
    print(f"MAE: {mae:.4f}")
    print(f"R-squared: {r2:.4f}")
    
    # Prediction vs Actual graph
    plt.figure(figsize=(10, 6))
    plt.scatter(y_test, y_pred, color='blue', alpha=0.6, label="Predictions")
    plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], color='red', linestyle='--', label="Ideal Fit")
    plt.title("Prediction vs Actual")
    plt.xlabel("Actual")
    plt.ylabel("Predicted")
    plt.legend()
    plt.show()

# ---- Before Tuning: Default Model ----

# Initialize LightGBM with default parameters
default_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', lgb.LGBMRegressor(random_state=42))
])

# Train the default model
default_model.fit(X_train, y_train)

# Evaluate the default model's performance
print("Performance Before Tuning:")
evaluate_model(default_model, X_train, y_train, X_test, y_test)

# ---- Bayesian Optimization for LightGBM ----

# Bayesian Optimization for LightGBM (with stricter regularization)
def lgb_eval(num_leaves, learning_rate, n_estimators, min_child_samples, subsample):
    params = {
        'num_leaves': int(num_leaves),
        'learning_rate': learning_rate,
        'n_estimators': int(n_estimators),
        'min_child_samples': int(min_child_samples),
        'subsample': subsample,
        'lambda_l1': 0.5,  # Increased L1 regularization
        'lambda_l2': 0.5,  # Increased L2 regularization
        'random_state': 42
    }
    
    model = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('regressor', lgb.LGBMRegressor(**params))
    ])
    
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    return -mean_absolute_error(y_test, pred)  # Negative because we want to maximize

bounds = {
    'num_leaves': (20, 100),
    'learning_rate': (0.01, 0.2),
    'n_estimators': (100, 500),
    'min_child_samples': (5, 50),
    'subsample': (0.6, 1.0)
}

optimizer = BayesianOptimization(f=lgb_eval, pbounds=bounds, random_state=42)
optimizer.maximize(init_points=5, n_iter=20)

# Get best parameters
best_params = optimizer.max['params']
best_params['num_leaves'] = int(best_params['num_leaves'])
best_params['n_estimators'] = int(best_params['n_estimators'])
best_params['min_child_samples'] = int(best_params['min_child_samples'])

# Final model with best parameters
final_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', lgb.LGBMRegressor(**best_params, random_state=42))
])

# Train the tuned model
final_model.fit(X_train, y_train)

# Evaluate the tuned model's performance
print("Performance After Tuning:")
evaluate_model(final_model, X_train, y_train, X_test, y_test)

```
